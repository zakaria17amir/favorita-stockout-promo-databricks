# 02 · Silver (Spark SQL)

Ported from Project 1's DuckDB SQL: `sales` (store × item × day, returns split out), `store_day` (receipts),
`national_holidays`, and `stockout_series`, the daily store-item series behind both stock-out tables.
Each `.sql` file in `src/favorita_spark/sql/silver/` is one SELECT; the runner wraps it in `CREATE OR REPLACE TABLE`.

```python
import os, sys
sys.path.insert(0, os.path.abspath("../src"))  # the repo's src/, via the Databricks Git folder

from favorita_spark.config import Config

cfg = Config()
spark.sql(f"USE CATALOG {cfg.catalog}")
```

```python
from favorita_spark.runner import build_sql_layer, render

build_sql_layer(spark, "silver", cfg)
print(render("silver", "stockout_series", cfg))
```

```python
%sql
SELECT 'sales' AS table_name, count(*) AS rows, min(date) AS first_day, max(date) AS last_day FROM silver.sales
UNION ALL SELECT 'store_day', count(*), min(date), max(date) FROM silver.store_day
UNION ALL SELECT 'stockout_series', count(*), min(date), max(date) FROM silver.stockout_series
```
